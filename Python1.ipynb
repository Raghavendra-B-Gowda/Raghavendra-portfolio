## Local and Global Variables

In Python, the scope of a variable refers to the region of the code where a variable is accessible.

*   **Global Variables**: Variables declared outside of any function are considered global variables. They can be accessed from anywhere in the program, both inside and outside functions.

*   **Local Variables**: Variables declared inside a function are local variables. They are only accessible within the function where they are defined. Once the function finishes execution, the local variables are destroyed.

```python
# Global variable
global_var = "I am a global variable"

def my_function():
    # Local variable
    local_var = "I am a local variable"
    print(f"Inside the function: {global_var}") # Accessing global_var
    print(f"Inside the function: {local_var}")  # Accessing local_var

my_function()

print(f"Outside the function: {global_var}") # Accessing global_var
# The following line would cause an error because local_var is not defined in the global scope:
# print(f"Outside the function: {local_var}")
```

## Method Overloading

Method overloading refers to having multiple methods with the same name but different parameters within the same class. Python does **not** support method overloading in the traditional sense (like Java or C++), where you can define two methods with the same name but different numbers or types of arguments.

However, you can achieve similar functionality using techniques like:

1.  **Default arguments**: Provide default values for parameters, making them optional.
2.  **Variable-length arguments (`*args`)**: Allow a function to accept an arbitrary number of positional arguments.
3.  **Keyword arguments (`**kwargs`)**: Allow a function to accept an arbitrary number of keyword arguments.

Let's demonstrate with default arguments, which is the most common way to handle varying arguments for a single function name in Python.

```python
class Calculator:
    def add(self, a, b=0, c=0):
        if b == 0 and c == 0:
            print(f"Adding one number: {a}")
            return a
        elif c == 0:
            print(f"Adding two numbers: {a} + {b}")
            return a + b
        else:
            print(f"Adding three numbers: {a} + {b} + {c}")
            return a + b + c

# Creating an instance of the Calculator class
calc = Calculator()

# Calling the add method with different numbers of arguments
print(f"Result (one argument): {calc.add(10)}")
print(f"Result (two arguments): {calc.add(10, 20)}")
print(f"Result (three arguments): {calc.add(10, 20, 30)}")
```

## Method Overriding

Method overriding occurs when a subclass (child class) provides its own specific implementation for a method that is already defined in its superclass (parent class). This allows a subclass to change the behavior of an inherited method, while keeping the same method signature (name, number of arguments).

When a method is called on an object, Python checks the object's class first. If the method is found there, it's executed. If not, Python looks up the inheritance chain to the parent class, and so on, until the method is found. This is how the overridden method in the subclass takes precedence.

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Create instances
animal = Animal()
dog = Dog()
cat = Cat()

# Call the speak method
animal.speak() # Output: Animal makes a sound
dog.speak()    # Output: Dog barks
cat.speak()    # Output: Cat meows
```

## Constructors (`__init__` method)

A constructor in Python is a special method called `__init__`. It is automatically called when you create a new object (instance) of a class. The primary purpose of a constructor is to initialize the attributes (variables) of the object.

It allows you to set up the initial state of the object right when it's created. The `self` parameter refers to the instance of the class itself.

# Inheritence (types)

Python supports several types of inheritance, allowing a class to inherit attributes and methods from one or more other classes. This promotes code reusability and helps establish a relationship between different classes.

Here are the main types of inheritance demonstrated:

1.  **Single Inheritance**:
    *   A class inherits from a single base class.
    *   Example: `Child` class inheriting from `Parent` class.

2.  **Multiple Inheritance**:
    *   A class inherits from multiple base classes.
    *   Example: `Child` class inheriting from both `Father` and `Mother` classes.

3.  **Multilevel Inheritance**:
    *   A class inherits from a base class, and then another class inherits from that derived class. It forms a chain of inheritance.
    *   Example: `Child` inherits from `Father`, and `Father` inherits from `Grandfather`.

4.  **Hierarchical Inheritance**:
    *   Multiple classes inherit from a single base class.
    *   Example: `Boys` and `Girls` classes both inheriting from the `Class` class.

5.  **Hybrid Inheritance**:
    *   A combination of two or more types of inheritance. It can involve a mix of single, multiple, or multilevel inheritance structures.
    *   Example: Class `D` inherits from `B` and `C`, while `B` and `C` both inherit from `A`.

```python
# single inheritence
class Parent:
    def show_parent(self):
        print("This is Parent")
class Child(Parent):
    def show_child(self):
        print("This is Child")
obj = Child()
obj.show_parent()
obj.show_child()
```

```python
#multiple inheritence
class Father:
    def father(self):
        print("Father")
class Mother:
    def mother(self):
        print("Mother")
class Child(Father, Mother):
    def child(self):
        print("Child")
obj = Child()
obj.father()
obj.mother()
obj.child()
```

```python
#multilevel inheritence
class Grandfather:
    def grandfather(self):
        print("Grandfather")
class Father(Grandfather):
    def father(self):
        print("Father")
class Child(Father):
    def child(self):
        print("Child")
obj = Child()
obj.grandfather()
obj.father()
obj.child()
```

```python
#hierarchical inheritance
class Class:
    def classroom(self):
        print("This is the class")
class Boys(Class):
    def boys(self):
        print("Boys are studying")
class Girls(Class):
    def girls(self):
        print("Girls are studying")
b = Boys()
b.classroom()
b.boys()
g = Girls()
g.classroom()
g.girls()
```

```python
#hybrid inheritence
class A:
    def show_a(self):
        print("A")
class B(A):
    def show_b(self):
        print("B")
class C(A):
    def show_c(self):
        print("C")
class D(B, C):
    def show_d(self):
        print("D")
obj = D()
obj.show_a()
obj.show_b()
obj.show_c()
obj.show_d()
```

# encapsulation

```python
class Emp:
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary >= 0:
            self.__salary = salary
        else:
            print("Salary cannot be negative")
# Create object
e1 = Emp(30000)
# Get salary
print(e1.get_salary())
# Change salary
e1.set_salary(40000)
# Get updated salary
print(e1.get_salary())
# Try negative salary
e1.set_salary(-5000)
```

```python
def DivideAmount(total, price):
    try:
        result = total / price

    except ZeroDivisionError:
        print("Zero Division Error")

    except TypeError:
        print("Type Error")

    else:
        print(result)

    finally:
        print("Finally")


DivideAmount(100, 0)
```

```python
import time
import numpy as np

# 1. Clear the overridden built-in if it exists in the current namespace
if 'list' in locals() or 'list' in globals():
    try:
        del list  # Removes your custom 'list' variable so Python uses the built-in function
    except NameError:
        pass

# Set the size of the collections
ELEMENT_COUNT = 1_000_000

print(f"Comparing performance using {ELEMENT_COUNT:,} elements...\n")

# --- PERFORMANCE TEST FOR PYTHON LIST ---
# Using list comprehension directly avoids relying on the list() constructor entirely
list_a = [i for i in range(ELEMENT_COUNT)]
list_b = [i for i in range(ELEMENT_COUNT)]

start_list_time = time.perf_counter()
list_result = [x + y for x, y in zip(list_a, list_b)]
end_list_time = time.perf_counter()

list_duration = end_list_time - start_list_time
print(f"Python List Execution Time: {list_duration:.6f} seconds")


# --- PERFORMANCE TEST FOR NUMPY ARRAY ---
array_a = np.arange(ELEMENT_COUNT)
array_b = np.arange(ELEMENT_COUNT)

start_array_time = time.perf_counter()
array_result = array_a + array_b
end_array_time = time.perf_counter()

array_duration = end_array_time - start_array_time
print(f"NumPy Array Execution Time: {array_duration:.6f} seconds")


# --- FINAL COMPARISON ---
speed_difference = list_duration / array_duration
print("-" * 45)
print(f"NumPy Array is roughly {speed_difference:.1f}x faster than a Python List.")
print("-" * 45)
```
